# BGE-M3 - Multilingual Hybrid Retrieval for 100 Languages on SageMaker

This notebook demonstrates how to subscribe to and use **BGE-M3 - Multilingual Hybrid Retrieval for 100 Languages on SageMaker** from AWS Marketplace, deploy it as a SageMaker real-time inference endpoint, and run predictions.

**Model:** `BAAI/bge-m3`  
**Price:** $0.10/hr (endpoint running time)

---


## Prerequisites

- AWS account with SageMaker access
- Subscribed to this product on [AWS Marketplace](https://aws.amazon.com/marketplace/)
- `boto3` and `sagemaker` Python packages installed

```bash
pip install boto3 sagemaker
```


````python
import boto3
import sagemaker
from sagemaker import ModelPackage

# Configuration
region = boto3.session.Session().region_name
role = sagemaker.get_execution_role()
print(f'Region: {region}')
print(f'Role: {role}')
````

## 1. Subscribe and Deploy

After subscribing on AWS Marketplace, retrieve the Model Package ARN for your region and deploy:


````python
# Replace with the Model Package ARN from your AWS Marketplace subscription
# AWS Marketplace Product ID: prod-vxfxjsxeid35i
model_package_arn = 'arn:aws:sagemaker:<REGION>:<ACCOUNT_ID>:model-package/<MODEL_PACKAGE_NAME>'

# Deploy the model
model = ModelPackage(
    role=role,
    model_package_arn=model_package_arn,
    sagemaker_session=sagemaker.Session()
)

predictor = model.deploy(
    initial_instance_count=1,
    instance_type='ml.m5.xlarge',
    endpoint_name=f'bge-m3-multilingual-hybrid-endpoint'
)

print(f'Endpoint deployed: {predictor.endpoint_name}')
````

## 2. Run Inference

Generate text embeddings. Returns a dense vector embedding.


````python
import json

# Generate text embeddings
payload = {"inputs": "The quick brown fox jumps over the lazy dog"}

response = predictor.predict(
    data=json.dumps(payload),
    initial_args={'ContentType': 'application/json'}
)

result = json.loads(response)
print('Result:', json.dumps(result, indent=2))
````

## 3. Clean Up

Delete the endpoint when finished to avoid charges:


````python
# Delete endpoint to stop billing
predictor.delete_endpoint()
print('Endpoint deleted')
````
