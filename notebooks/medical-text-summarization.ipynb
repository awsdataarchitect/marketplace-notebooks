## Step 1: Subscribe on AWS Marketplace

Before deploying, subscribe to this model on AWS Marketplace. After subscribing, the Model Package will appear in your SageMaker console.

**[Subscribe on AWS Marketplace →](#)**


```python
import boto3
import sagemaker
import json

# Setup
region = boto3.session.Session().region_name
role = sagemaker.get_execution_role()
sess = sagemaker.Session()

print(f'Region: {region}')
print(f'SageMaker Role: {role}')
```

```python
from sagemaker import ModelPackage

# After subscribing, paste your Model Package ARN here
# AWS Marketplace Product ID: prod-ploqu6fg6arda
model_package_arn = 'arn:aws:sagemaker:REGION:ACCOUNT:model-package/MODEL_PACKAGE_NAME'

# Validate ARN before deploying
if 'REGION' in model_package_arn or 'ACCOUNT' in model_package_arn or 'MODEL_PACKAGE_NAME' in model_package_arn:
    raise ValueError(
        'Please replace REGION, ACCOUNT, and MODEL_PACKAGE_NAME with actual values '
        'from your AWS Marketplace subscription receipt email.'
    )

# Deploy the endpoint
model = ModelPackage(
    role=role,
    model_package_arn=model_package_arn,
    sagemaker_session=sess
)

predictor = model.deploy(
    initial_instance_count=1,
    instance_type='ml.m5.xlarge',
    endpoint_name='medical-text-summarization-demo'
)

print(f'Endpoint ready: {predictor.endpoint_name}')
```

## Step 2: Run Inference

Summarize medical text:


```python
# Summarize medical text
payload = {"inputs": "Patient presents with [symptom description] and [vital signs]. Diagnostic evaluation reveals [test findings]. Assessment: [diagnosis]. Plan: [treatment protocol] initiated per [clinical guideline]."}

response = predictor.predict(
    data=json.dumps(payload),
    initial_args={'ContentType': 'application/json'}
)

try:
    result = json.loads(response)
    print(json.dumps(result, indent=2))
except json.JSONDecodeError as e:
    print(f"Error parsing response: {e}")
    print(f"Raw response: {response}")
```

## Step 3: Clean Up

Delete the endpoint to stop billing:


```python
predictor.delete_endpoint()
print('Endpoint deleted -- billing stopped')
```
