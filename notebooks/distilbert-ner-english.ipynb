```python
import json

import boto3
import sagemaker
from sagemaker import ModelPackage

# Paste the model package ARN from the product's launch page for YOUR Region.
MODEL_PACKAGE_ARN = "<paste-model-package-arn-here>"

INSTANCE_TYPE = "ml.m5.xlarge"  # recommended real-time instance
ENDPOINT_NAME = "distilbert-ner-english"

session = sagemaker.Session()
role = sagemaker.get_execution_role()
print("region:", session.boto_region_name)
```

## 2. Deploy a real-time endpoint

Takes roughly 6-9 minutes. The endpoint bills per hour while it exists, so do not skip section 5.

```python
model = ModelPackage(
    role=role,
    model_package_arn=MODEL_PACKAGE_ARN,
    sagemaker_session=session,
)

predictor = model.deploy(
    initial_instance_count=1,
    instance_type=INSTANCE_TYPE,
    endpoint_name=ENDPOINT_NAME,
)
print("endpoint ready:", ENDPOINT_NAME)
```

## 3. Extract entities

The endpoint accepts `application/json` shaped `{"inputs": "..."}`.
Returns a list of entities: `[{"entity": "ORG", "word": "Apple Inc.", "score": 0.99, "start": 0, "end": 10}, ...]`

Entity types: PER (person), ORG (organisation), LOC (location), MISC (miscellaneous). Scores indicate per-token confidence.

```python
runtime = boto3.client("sagemaker-runtime")


def extract_entities(text):
    """Return named entities found in the input text."""
    response = runtime.invoke_endpoint(
        EndpointName=ENDPOINT_NAME,
        ContentType="application/json",
        Body=json.dumps({"inputs": text}),
    )
    return json.loads(response["Body"].read())


text = "Apple Inc. CEO Tim Cook announced new products at the Cupertino headquarters."
entities = extract_entities(text)

for ent in entities:
    print(f"{ent.get('entity', ent.get('entity_group', '?')):12s} "
          f"score={ent.get('score', 0):.3f}  word={ent.get('word', '')}")
```
